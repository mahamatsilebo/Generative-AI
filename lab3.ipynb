```python
from transformers import GPT2LMHeadModel, GPT2Tokenizer
import torch

# Load GPT-2
tokenizer = GPT2Tokenizer.from_pretrained("gpt2")
model = GPT2LMHeadModel.from_pretrained("gpt2")

# Use GPU if available
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model = model.to(device)

print("Model loaded successfully!")
print("Device:", device)
```

```python
prompt = "Artificial intelligence is changing the way people work"
```

```python
inputs = tokenizer(prompt, return_tensors="pt").to(device)

with torch.no_grad():
    output = model.generate(
        **inputs,
        max_length=40,
        do_sample=True,
        temperature=1.0,
        top_k=0,
        top_p=1.0
    )

print(tokenizer.decode(output[0], skip_special_tokens=True))
```

```python
def generate_text(prompt, temperature=1.0, top_k=0, top_p=1.0):
    inputs = tokenizer(prompt, return_tensors="pt").to(device)

    with torch.no_grad():
        output = model.generate(
            **inputs,
            max_length=40,
            do_sample=True,
            temperature=temperature,
            top_k=top_k,
            top_p=top_p
        )

    return tokenizer.decode(output[0], skip_special_tokens=True)
```

```python
# 1. Very Low Temperature
text1 = generate_text(
    prompt,
    temperature=0.1,
    top_k=0,
    top_p=1.0
)

# 2. Very High Temperature
text2 = generate_text(
    prompt,
    temperature=1.8,
    top_k=0,
    top_p=1.0
)

# 3. Narrow Top-K
text3 = generate_text(
    prompt,
    temperature=1.0,
    top_k=2,
    top_p=1.0
)

# 4. Wide Top-K
text4 = generate_text(
    prompt,
    temperature=1.0,
    top_k=50,
    top_p=1.0
)

# 5. Strict Top-P
text5 = generate_text(
    prompt,
    temperature=1.0,
    top_k=0,
    top_p=0.5
)

# 6. Loose Top-P
text6 = generate_text(
    prompt,
    temperature=1.0,
    top_k=0,
    top_p=0.95
)
```

```python
print("=" * 70)
print("1. VERY LOW TEMPERATURE (0.1)")
print("=" * 70)
print(text1)

print("\n" + "=" * 70)
print("2. VERY HIGH TEMPERATURE (1.8)")
print("=" * 70)
print(text2)

print("\n" + "=" * 70)
print("3. NARROW TOP-K (2)")
print("=" * 70)
print(text3)

print("\n" + "=" * 70)
print("4. WIDE TOP-K (50)")
print("=" * 70)
print(text4)

print("\n" + "=" * 70)
print("5. STRICT TOP-P (0.5)")
print("=" * 70)
print(text5)

print("\n" + "=" * 70)
print("6. LOOSE TOP-P (0.95)")
print("=" * 70)
print(text6)
```

The experiment demonstrates that text-generation parameters have a significant effect on the behavior of GPT-2. Lower temperature and restrictive sampling methods generally produce more predictable and focused text, while higher temperature and less restrictive sampling allow greater diversity and creativity. However, excessive randomness can reduce coherence. Therefore, temperature, top-k, and top-p can be adjusted depending on whether the desired output prioritizes consistency, creativity, or a balance between the two.
