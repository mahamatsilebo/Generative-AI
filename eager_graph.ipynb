```python
import tensorflow as tf

# Various text inputs
texts = [
    "Hello World",
    "TensorFlow is easy to learn",
    "I am studying Data Science",
    "Machine Learning is powerful",
    "Natural Language Processing is interesting"
]

# =========================================================
# 1. EAGER MODE
# =========================================================

print("========== EAGER MODE ==========")
print("Eager execution:", tf.executing_eagerly())

for text in texts:
    tensor = tf.constant(text)

    # Text operations
    uppercase = tf.strings.upper(tensor)
    length = tf.strings.length(tensor)

    print("\nOriginal :", text)
    print("Uppercase:", uppercase.numpy().decode())
    print("Length   :", length.numpy())

```

```python

# =========================================================
# 2. GRAPH MODE
# =========================================================

print("\n========== GRAPH MODE ==========")

@tf.function
def process_text(text):
    uppercase = tf.strings.upper(text)
    length = tf.strings.length(text)

    return uppercase, length


for text in texts:
    tensor = tf.constant(text)

    uppercase, length = process_text(tensor)

    print("\nOriginal :", text)
    print("Uppercase:", uppercase.numpy().decode())
    print("Length   :", length.numpy())
```
